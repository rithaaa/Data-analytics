Creating data frame 

```python
import pandas as pd
df=pd.DataFrame({
    'Name':['anu','hiba','jiya'],
    'Age': [21,22,13],
    'Mark': [12,34,90]
})
print(df)
```

To access or select a column

```python
print(df['Name'])
print("")
print(df['Age'])
print("")
print(df['Mark'])
```

To select Multiple Columns

```python
print(df[['Name','Mark']])
print(df[['Name','Age','Mark']])
```

To Add a new column

```python
df['Grade']=['A','A+','A+']
print(df)
```

Create a column using calculation

```python
df['Bonus Mark']=df['Mark']+5
print(df)
```

TO DROP A COLUMN

```python
df=df.drop('Bonus Mark',axis=1)
print(df)
```

```python
# Create a dataframe.

import pandas as pd
df=pd.DataFrame({
    'Mark1':[21,21],
    'Mark2':[34,56]
})
print(df)
```

```python
df=pd.DataFrame({
    'Name':[1,2,3],
    'Age':[23,56,78],
    'Mark':[12,54,32]
})
print(df)
print("")
df['Age']=df['Age']+10
print(df)
print("")
df['Mark']=df['Mark']+5
print(df)
```

TO GIVE NAME TO ROW

```python
df=pd.DataFrame({
    'Name':[1,2,3],
    'Age':[23,56,78],
    'Mark':[12,54,32]
},index=['a','b','c'])
print(df)
```

**STATISTICAL FUNCTIONS IN PANDAS**

.sum()
.median()
.max()
.min()
.std()
.count()

```python
print(df['Age'].sum())
print(df['Mark'].median())
print(df['Mark'].max())
print(df['Mark'].min())
print(df['Mark'].std())
print(df['Mark'].count())   #Gives the number of elements/values in the column.
```
