```python
# create array of size 3x3 and perform temperature adjustment with the following data.
# 1.Create Array
# 2.create temp adjustment
# 3.multiply a1 with temp adjustment
# 4.Add each column

a1=np.array([[42,34,45],[23,32,12],[89,67,45]])
print(a1)
print("")

temp_adj=np.array([[0.5,-1.5,2]])
print(temp_adj)
print("")

res=a1*temp_adj
print(res)
print("")

res1=np.sum(res,axis=0)
print(res1)
```

```python
# Create an array of 3 products over 4 months enter its sales each month calculate their highest month wise product sales.
# 1. find the maximum product sold.
# 2. find the avg number of product sold in total.

import numpy as np
ar=np.array([[20,30,40,50],[160,70,80,120],[200,67,125,78]])
maximum=np.max(ar,axis=0)
print(maximum)
print("")
maxi=np.max(ar,axis=1)
print(maxi)
print("")
av=np.avg(ar,axis=1)
print(av)
```
