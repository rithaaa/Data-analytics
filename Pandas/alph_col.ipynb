```python
# create a dataframe of 7 rows and 4 columns are 'name','dept('hr','it'),'college','city'.

import pandas as pd
df=pd.DataFrame({
    'name': ['A','B','C','d','e','F','g'],
    'dept': ['hr','it','hr','hr','it','it','hr'],
    'college': ['ABV','CVD','KJH','QWE','OUT','ABV','KJH'],
    'city': ['texas','chicago','new york','texas','chicago','texas','texas']
})
df
```

```python
df.groupby('dept')['city'].agg(
    lambda x: x.str.upper()
)
```
