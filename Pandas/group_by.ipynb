```python
# Create a dataframe of 7 rows and 6 columns.columns are name,dept,mobile_no,working_hrs,city,salary.

import pandas as pd
df=pd.DataFrame({
    'st_name': ['anu','hiba','aju','niju','chikku','naja','ritha'],
    'dept':['cse','eee','it','hr','cse','it','eee'],
    'mobile_no':[1234,4321,6789,9876,10934,67234,78342],
    'working_hrs': [9,12,5,7,3,9,5],
    'city': ['kochi','hyderabad','chennai','mumbai','banglore','kochi','hyderabad'],
    'salary': [10000,20000,50000,5000,78000,7800,3000]
})
print(df)
print("")
print(df.groupby('city'))
print("")
print(df.groupby('dept')['salary'])
```

```python
print(df.groupby('dept')['salary'].mean())
```

```python
print(df.groupby('dept')['salary'].max())
```

```python
print(df.groupby('dept')['salary'].min())
```

```python
print(df.groupby('dept')['salary'].sum())
```

```python
df.groupby('dept')['salary'].agg(['max','min','sum'])
```

DIFFERENT COLUMN, DIFFERENT FUNCTION


```python
df.agg({
    'salary':'max',
    'working_hrs':'min'
})
```

```python
df.groupby('dept').agg({
    'salary': 'max',
    'working_hrs': 'min'
})
```

different function on the same column

```python
df.agg(
    max_sal=('salary','max'),  
    mean_sal=('salary','mean')
)
```

SORT VALUES 

```python
result=df.groupby('dept')['salary'].max()
print(result.sort_values(ascending=True))
print("")
print(result.sort_values(ascending=False))
```

```python
# Create a dataframe Name,City,Occupation,clg_name,dept,salary.
# 1.group by city and find minimum salary.
# 2.group by names and sort it by city.
# 3.Group by name and convert their occupation into uppercase.

import pandas as pd
df=pd.DataFrame({
    'name': ['anu','hiba','aju','chikku','chinju'],
    'city': ['kochi','calicut','hyderabad','kochi','hyderabad'],
    'occupation':['driver','teacher','teacher','gardener','driver'],
    'clg_name':['abc','xyz','abc','cvd','cvd'],
    'dept': ['it','hr','it','mech','mech'],
})
df
```

```python
# 1.group by city and find minimum salary.

df['salary']=[20000,40000,40000,25000,20000]
print(df)
print("")
df.groupby('city')['salary'].min()
```

```python
# 2.group by names and sort it by city.


a=df.groupby('name')['city'].agg(
    lambda x: x.sort_values(ascending=True)
)
a
```

```python
# 3.Group by name and convert their occupation into uppercase.

df.groupby('name')['occupation'].agg(
    lambda x: x.str.upper()
)
```
