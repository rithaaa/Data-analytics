```python
# Create a DataFrame of 4x3, column with values city,population,no_of_socail_welfare_worker.

import pandas as pd
df=pd.DataFrame({
    'city': ['Kochi','Chennai',None,'Kochi'],
    'population': [123,564,347,367],
    'no_of_socail_welfare_worker': [34,57,31,12]
})
print(df)
```

```python
# To count Unique values.

df['city'].value_counts()
```

```python
# To find the percentage

print(df['population'].value_counts(normalize=True))
print("")
print(df['population'].value_counts(normalize=True)*100)
```

```python
# To Include missing values

df['city'].value_counts(dropna=False)
```

```python
# No of unique values

df['city'].nunique()
```

```python
# To get unique values

print(df['city'].unique())
```

```python
# Create 2 dataframe 

a=pd.DataFrame({
    'name': ['Arathy','Neha'],
    'age': [13,22]
})
print(a)
print("")
b=pd.DataFrame({
    'maths': [30,27],
    'science': [26,20]
})
print(b)
print("")
print(pd.concat([a,b]))
print("")
print(pd.concat([a,b],ignore_index=True))
print("")
print(pd.concat([a,b],axis=1))
```

```python
# Create 2 dataframe and merge based on their ID.

c=pd.DataFrame({
    'ID': [0,1],
    'name': ['Arathy','Neha']
})
print(c)
print("")
d=pd.DataFrame({
    'ID': [1,2],
    'mark': [26,20]
})
print(d)
print("")
pd.merge(c,d,on="ID")
```

```python
# Create a dataframe of student with their details of name,age,father's_name,mother's_name,mobile_no and their Id.
# Create a student details with st_id,st_name,class,test1,test2,mobile_no.
# 5 rows

st_p=pd.DataFrame({
    'st_name':['anu','hiba','neha','ichu','niju'],
    'age':[21,23,45,67,43],
    'father_name': ['biju','shiju','unni','hari','shareef'],
    'mother_name':['bindu','sindhu','sheeba','shyma','rajula'],
    'mobile_no': [1234,4321,56789,98765,1230],
    'st_id': [101,102,103,104,105]
})
print(st_p)
st_d=pd.DataFrame({
    'st_id': [105,104,110,102,101],
    'st_name':['niju','ichu','neha','hiba','anu'],
    'class':['cse','ece','cse','mech','eee'],
    'test1': [12,21,20,17,15],
    'test2': [11,10,8,15,16],
    'mobile_no': [1230,1672,56789,4321,1234],
})
print("")
print(st_d)
print("")
```

```python
pd.merge(st_p,st_d,on='st_id')
```

```python
pd.merge(st_p,st_d,on='st_name')
```

```python
pd.merge(st_p,st_d,on='mobile_no')
```

```python
pd.merge(st_p,st_d,on='st_id',how='left')
```

```python
pd.merge(st_p,st_d,on='st_id',how='right')
```

```python
pd.merge(st_p,st_d,on='st_id',how='inner')
```

```python
pd.merge(st_p,st_d,on='st_id',how='outer')
```
