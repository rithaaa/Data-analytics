```python
# Create a dataframe of 4x1, name the columns as marks.

import pandas as pd
df=pd.DataFrame({
    'Marks': [23,45,65,12]
}, index=['s1','s2','s3','s4'])
print(f"Original: \n",df)
print('')
print(f"Default Sort:\n ",df.sort_values('Marks'))
print('')
print(f"Ascending:\n ",df.sort_values('Marks',ascending=True))
print('')
print(f"Descending:\n ",df.sort_values('Marks',ascending=False))
print("")
print(f"First value:\n ",df.head(1))
print("")
print(df.loc['s4',:])
```

```python
df=df.sort_values('Marks',ascending=False)
print(df.tail(1))
```

```python
print(df['Marks']>50)
print("------------------------")
df[df['Marks']>50]
print("------------------------")
```
