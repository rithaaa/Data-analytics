```python
import seaborn as sns
import pandas as pd
```

corr --- is used to find the co-relation between the variable.

```python
df=sns.load_dataset('tips')
sns.heatmap(
    df.corr(numeric_only=True)
)
```

```python
df=sns.load_dataset('tips')
sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=1,
    cbar=True
)
```

```python
import seaborn as sns
df =sns.load_dataset('tips')
sns.kdeplot(
    data=df,
    x="total_bill"
)
```

```python
sns.kdeplot(
    data=df,
    x="total_bill",
    fill=True
)
```

```python
sns.kdeplot(
    data=df,
    x="size",
    color='red'
)
```

```python
sns.kdeplot(
    data=df,
    x="size",
    fill=True,
    cut=0
)
```

```python
sns.kdeplot(
    data=df,
    x="size",
    fill=True,
    cut=0,
    hue="sex"
)
```

```python
sns.kdeplot(
    data=df,
    x="size",
    fill=True,
    cut=0,
    hue="sex",
    multiple="stack"
)
```

```python
sns.kdeplot(
    data=df,
    x="size",
    fill=True,
    cut=0,
    hue="sex",
    multiple="layer"
)
```

```python
sns.kdeplot(
    data=df,
    x="size",
    fill=True,
    cut=0,
    hue="sex",
    multiple="fill"
)
```

```python
sns.kdeplot(
    data=df,
    x="size",
    fill=True,
    cut=0,
    hue="sex",
    multiple="stack",
    palette="Set1"          #palette="Set1" "Set2"  "Set3"  "pastel" "bright" "dark"  "coolwarm"
)
```

```python
sns.kdeplot(
    data=df,
    x="size",
    fill=True,
    cut=0,
    hue="sex",
    multiple="stack",
    palette={
             "Male":"blue",
             "Female":"pink"
            }
)
```
