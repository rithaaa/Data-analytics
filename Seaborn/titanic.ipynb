```python
import seaborn as sns
tf=sns.load_dataset('titanic')
tf
```

```python
sns.heatmap(
    tf.corr(numeric_only=True)
)
```

```python
sns.heatmap(
    tf.corr(numeric_only=True),
    annot=False,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=.5,
    cbar=True
)
```

```python
sns.heatmap(
    tf.corr(numeric_only=True),
    annot=False,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=1,
    cbar=True,
    linecolor='orange'
)
```

```python
sns.heatmap(
    tf.corr(numeric_only=True),
    annot=False,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=1,
    cbar=True,
    vmax=1,
    vmin=1
)
```

```python
sns.heatmap(
    tf.corr(numeric_only=True),
    annot=False,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=1,
    cbar=False,
    center=False,
    square=True,
    xticklabels=False
)
```

```python
sns.kdeplot(
    data=tf,
    x="survived",
    fill=True,
    color="red",
    cut=0
)
```

```python
sns.kdeplot(
    data=tf,
    x="survived",
    fill=True,
    color="red",
    cut=0,
    hue="age",
    multiple="stack",
    palette="coolwarm"  
)
```
