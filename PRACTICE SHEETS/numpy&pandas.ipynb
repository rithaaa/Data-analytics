A1 — DATAFRAME CREATION & COLUMN OPERATIONS

```python
# 1. Create a DataFrame containing Name, Marks, and Age for 8 students. Create a new column Result based on whether the student's marks satisfy a condition you decide.

import pandas as pd
df=pd.DataFrame({
    'Name': ['anu','hiba','naja','aishu','ritha','niju','chakki','chikku'],
    'Marks': [34,45,21,12,45,50,25,46],
    'Age': [21,22,25,46,31,22,18,21]
})
print(df)
print("")
df['Result']=df['Marks'].apply(
    lambda x:("Pass") if x>=25 else ('Fail')
)
print(df)
```

```python
# 2. Create a DataFrame containing First_Name and Last_Name for 7 people. Create a Full_Name column by combining both columns.

import pandas as pd
df=pd.DataFrame({
    'First_Name': ['Ritha','Hiba','Arun','Hari','Vijay','Sathya','Niyas'],
    'Last_Name': ['Fathima','Basheer','Kumar','Kumar','Yeshudas','Seelan','Kala']
})
print(df)
print("")

df['Full_Name']=df['First_Name']+' '+df['Last_Name']
df
```

```python
# 3. Create an employee DataFrame with Name, Salary, and Bonus. Create a new column containing the employee's total income.

import pandas as pd
employee=pd.DataFrame({
    'Name': ['anu','hiba','naja','aishu','ritha'],
    'Salary': [1000,2000,1000,3500,1000],
    'Bonus': [20,10,5,20,10]
})
print(employee)
print("")
employee['total income']= employee['Salary']+employee['Bonus']
print(employee)
```

```python
# 4. Create a DataFrame with Product, Price, and Quantity. Create a new column containing the total price for each product.

import pandas as pd
detail=pd.DataFrame({
    'Product': ['mobile','laptop','TV','mobile','AC'],
    'Price': [10000,20000,15000,25000,20000],
    'Quantity': [2,5,6,1,2]
})
print(detail)
print("")
detail['total price']=detail['Price'] * detail['Quantity']
print(detail)
```

```python
# 5. Create a student DataFrame with Name, Maths, and Science. Create a new column containing the total marks.

import pandas as pd
student=pd.DataFrame({
    'Name': ['Ritha','Chaithra','Sanika','Abhirami','Surya'],
    'Maths': [45,23,37,48,50],
    'Science': [25,50,36,41,29]
})
print(student)
print("")
student['total_marks']=student['Maths']+student['Science']
print(student)
```

```python
# 6. Using the same DataFrame, create another column containing the average marks.

student['avg_mark']=student['total_marks']/2
print(student)
```

```python
# 7. Create an employee DataFrame with Name, Salary, and Experience. Create a new column using both salary and experience.

import pandas as pd
employee=pd.DataFrame({
    'Name': ['Ritha','Chaithra','Sanika','Abhirami','Surya'],
    'Salary': [45000,23000,37000,48000,50000],
    'Experience': [2,5,6,1,9]
})
print(employee)
print("")
employee['Bonus_salary']=(employee['Salary']*employee['Experience']/100)
print(employee)
```

```python
# 8. Create a DataFrame containing Name, Age, City, and Salary. Change all values in the City column to another set of cities.

import pandas as pd
person=pd.DataFrame({
    'Name': ['Ritha','Chaithra','Sanika','Abhirami','Surya'],
    'Age': [25,23,37,48,50],
    'City': ['calicut','cochin','palakkad','thrissur','kannur'],
    'Salary': [45000,23000,37000,48000,50000],
})
print(person)
print("")
person['City']=['new york','india','kochi','america','dubai']
print(person)
```

```python
# 10. Create a DataFrame containing Name, Department, and Salary. Create a new column that represents salary after applying a bonus percentage.

import pandas as pd
employee=pd.DataFrame({
    'Name': ['Ritha','Chaithra','Sanika','Abhirami','Surya'],
    'Department': ['cs','ec','ee','cs','ee'],
    'Salary': [45000,23000,37000,48000,50000]
})
print(employee)
print("")
employee['Bonus_salary']=employee['Salary']+(employee['Salary']*0.25)
print(employee)
```

```python
# 18. Create an employee DataFrame and update salaries for employees whose current salary satisfies a condition.

import pandas as pd
employee = pd.DataFrame({
    'Name': ['Ritha', 'Hiba', 'Niju', 'Sanika', 'Arun'],
    'Department': ['CS', 'HR', 'CS', 'IT', 'HR'],
    'Salary': [25000, 35000, 28000, 50000, 30000]
})
print(employee)
print('')
employee.loc[employee['Salary'] < 30000, 'Salary'] += 5000
print(employee)
```

```python
# 19. Change values in two different columns for a selected row.

employee = pd.DataFrame({
    'Name': ['Ritha', 'Hiba', 'Niju', 'Sanika'],
    'Department': ['CS', 'HR', 'IT', 'CS'],
    'Salary': [25000, 30000, 35000, 40000]
}, index=[101, 102, 103, 104])
print(employee)
print('')
employee.loc[102, ['Department', 'Salary']] = ['CS', 30000]
print(employee)
```

```python
# 20. Create a DataFrame and update multiple rows and multiple columns in one operation.


employee = pd.DataFrame({
    'Name': ['Ritha', 'Hiba', 'Niju', 'Sanika'],
    'Department': ['CS', 'HR', 'IT', 'CS'],
    'Salary': [25000, 30000, 35000, 40000]
}, index=[101, 102, 103, 104])
print(employee)
print("")
employee.loc[[101, 103], ['Department', 'Salary']] = [
    ['IT', 30000],
    ['HR', 40000]
]
print(employee)
```
